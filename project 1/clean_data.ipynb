```python
import pandas as pd
```

```python
# import csv file

df = pd.read_csv(r"C:\Users\lenovo\Downloads\Amazon_laptops.csv")
df
```

```python
# 1. convert title in correct form

df.columns = df.columns.str.strip().str.title()
df.columns
```

```python
# 2. see the first 5 rows
df.head()
```

```python
# 3.check the basic information
df.info()
```

```python
# 4. check duplicate value
df.duplicated().sum()
```

```python
# 5.remove duplicate row
df = df.drop_duplicates()
df.duplicated().sum()
```

```python
# 6. check missing values
df.isnull().sum()
```

```python
# 8. remove extra spaces from text names
for col in df.select_dtypes(include="object"):
    df[col]=df[col].str.strip()

df[col]
```

```python
for col in df.select_dtypes(include="object"):
    df[col]=df[col].fillna("Unknown")

df[col]
```

```python
# final check
print(df.info())
print()
print(df.isnull().sum())
print()
print(df.duplicated().sum())
```

```python
df.shape
```

```python
missing_percentage = (df.isnull().sum() / len(df)) * 100

print(missing_percentage)
```

```python
df['Price']=df['Price'].fillna( df['Price'].median())
df['Price']
```

```python
# Converting from text to numeric, removing ₹.
# Converting text prices to numeric
df["Price"] = (
    df["Price"]
    .astype(str)
    .str.replace("₹", "", regex=False)
    .str.replace(",", "", regex=False)
    .str.strip()
)

df["Price"] = pd.to_numeric(df["Price"], errors="coerce")
df["Price"]
```

```python
# rating

df["Rating"].dtype
df["Rating"].unique()
```

```python
import numpy as np
df.loc[(df['Rating']<0) | (df['Rating']>5),
"Rating"] =np.nan
```

```python
df['Rating']=df['Rating'].fillna(df['Rating'].median())
```

```python
print(df['Rating'].isnull().sum())
```

```python
# clean brand
df["Brand"]=df["Brand"].str.strip().str.title()
df["Brand"]
```

```python
print(df["Brand"].value_counts())
```

```python
# Extract RAM number, e.g. 16GB → 16 and 1TB to 1064
ram = df["Ram"].astype(str)

df["Ram"] = ram.str.extract(r"(\d+\.?\d*)")[0].astype(float)

df.loc[ram.str.contains("TB", case=False, na=False), "Ram"] *= 1064

df["Ram"] = df["Ram"].bfill().ffill()

df["Ram"]
```

```python
# Extract storage number and normalize TB values to GB
Storage = df["Storage"].astype("string").str.upper().str.strip()

df["Storage"] = Storage.str.extract(r"(\d+\.?\d*)")[0].astype(float)

df.loc[Storage.str.contains("TB", case=False, na=False), "Storage"] *= 1024

df["Storage"] = df["Storage"].bfill().ffill()

df["Storage"]
```

```python
df.to_csv("laptops_data_cleaned.csv", index=False)
```
