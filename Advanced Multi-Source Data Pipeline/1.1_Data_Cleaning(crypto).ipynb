```python
# 1. MISSING VALUES
# Intelligent filling using median

numeric_cols = [
    "current_price",
    "market_cap",
    "fully_diluted_valuation",
    "total_volume",
    "high_24h",
    "low_24h",
    "price_change_24h",
    "price_change_percentage_24h",
    "market_cap_change_24h",
    "circulating_supply",
    "total_supply",
    "max_supply"
]

for col in numeric_cols:
    if col in df.columns:
        df[col] = pd.to_numeric(df[col], errors="coerce")

# Fill normal numeric missing values with median
for col in df.select_dtypes(include="number").columns:
    df[col] = df[col].fillna(df[col].median())

df[numeric_cols].isna().sum()
```

```python
# 2. INCORRECT DATA TYPES
# Convert price, market_cap and volume to float

for col in ["current_price", "market_cap", "total_volume"]:
    if col in df.columns:
        df[col] = pd.to_numeric(df[col], errors="coerce").astype(float)

df.dtypes[["current_price", "market_cap", "total_volume"]] 
```

```python
# 3. EXTREME OUTLIERS - IQR

outlier_cols = [
    "current_price",
    "market_cap",
    "total_volume"
]

df["IQR_Outlier"] = False

for col in outlier_cols:
    if col in df.columns:

        Q1 = df[col].quantile(0.25)
        Q3 = df[col].quantile(0.75)

        IQR = Q3 - Q1

        lower_limit = Q1 - 1.5 * IQR
        upper_limit = Q3 + 1.5 * IQR

        df.loc[
            (df[col] < lower_limit) |
            (df[col] > upper_limit),
            "IQR_Outlier"
        ] = True

df["IQR_Outlier"].value_counts()
```

```python
# 4. EXTREME OUTLIERS - ISOLATION FOREST
features = [
    "current_price",
    "market_cap",
    
    "total_volume"
]

model = IsolationForest(
    contamination=0.05,
    random_state=42
)

df["Isolation_Forest_Outlier"] = (
    model.fit_predict(df[features]) == -1
)
df["Isolation_Forest_Outlier"].value_counts()
```

```python
# 5. INCONSISTENT NULLS
# max_supply and total_supply handling

# max_supply can legitimately be unavailable
# so keep missing max_supply as NaN

# total_supply: fill missing values using circulating supply
if "total_supply" in df.columns and "circulating_supply" in df.columns:

    df["total_supply"] = df["total_supply"].fillna(
        df["circulating_supply"]
    )

df[["total_supply", "circulating_supply"]].isna().sum()
```

```python
# 6. DATE FORMAT
# Convert last_updated to datetime


if "last_updated" in df.columns:

    df["last_updated"] = pd.to_datetime(
        df["last_updated"],
        errors="coerce"
    )

df["last_updated"].dtypes
```

```python
# 7. DUPLICATE COINS
# Remove duplicates based on ID or symbol


if "id" in df.columns:

    df = df.drop_duplicates(
        subset="id",
        keep="first"
    )

elif "symbol" in df.columns:

    df = df.drop_duplicates(
        subset="symbol",
        keep="first"
    )

df.duplicated().sum()
```

```python
# 8. NEGATIVE OR ZERO PRICES
# Filter invalid records


if "current_price" in df.columns:

    df = df[
        df["current_price"] > 0
    ]

df["current_price"].min()
```

```python
# 9. PERCENTAGE COLUMNS
# Handle missing % change values


percentage_cols = [
    col for col in df.columns
    if "percentage" in col.lower()
]

for col in percentage_cols:

    df[col] = pd.to_numeric(
        df[col],
        errors="coerce"
    )

    df[col] = df[col].fillna(0)

df[percentage_cols].isna().sum()
```

```python
import numpy as np
# 10. FEATURE ENGINEERING
# Market Cap Category


if "market_cap" in df.columns:

    df["Market_Cap_Category"] = pd.cut(
        df["market_cap"],
        bins=[
            -np.inf,
            100_000_000,
            1_000_000_000,
            10_000_000_000,
            np.inf
        ],
        labels=[
            "Small",
            "Medium",
            "Large",
            "Mega"
        ]
    )

df["Market_Cap_Category"].value_counts()
```

```python
# 11. FEATURE ENGINEERING
# Volatility Score


if "total_volume" in df.columns:

    df["Volatility_Score"] = (
        df["total_volume"] /
        df["total_volume"].median()
    )

df["Volatility_Score"].describe()
```

```python
# 12. FEATURE ENGINEERING
# Supply Ratio


if (
    "circulating_supply" in df.columns
    and "total_supply" in df.columns
):

    df["Supply_Ratio"] = (
        df["circulating_supply"] /
        df["total_supply"].replace(0, np.nan)
    )

    df["Supply_Ratio"] = df["Supply_Ratio"].clip(
        lower=0,
        upper=1
    )

df["Supply_Ratio"].describe()
```

```python

# 13. FINAL CLEANING


df = df.replace(
    [np.inf, -np.inf],
    np.nan
)

```
